```python
# Define directories
input_directory = r"D:\Iniciação Científica\untreated_images\images\Entry_PNG_copy_2"
output_directory = r"D:\Iniciação Científica\untreated_images\images\Entry_PNG_cropped"

# Create output directory if it doesn't exist
os.makedirs(output_directory, exist_ok=True)

# Verify input directory exists
if not os.path.isdir(input_directory):
    print(f"Error: Input directory does not exist: {input_directory}")
    exit()

# Get a list of PNG image files in the directory
image_files = glob.glob(os.path.join(input_directory, "*.png"))

# Check if there are any images
if len(image_files) == 0:
    print(f"Error: No PNG images found in the directory: {input_directory}")
    exit()
else:
    print(f"Found {len(image_files)} PNG images in the directory.")

# Select 5 random images for display (or all if less than 5)
display_images = random.sample(image_files, min(5, len(image_files)))

# Process all images
for img_path in image_files:
    try:
        # Verify file exists
        if not os.path.isfile(img_path):
            print(f"Error: File does not exist: {img_path}")
            continue

        # Load image with PIL
        imagem = Image.open(img_path)
        print(f"Loaded image: {os.path.basename(img_path)}")

        # Display original image if it's one of the 5 selected
        if img_path in display_images:
            print(f"Displaying original image: {os.path.basename(img_path)}")
            display(imagem)

        # Get image dimensions
        width, height = imagem.size

        # Calculate crop height (30% of the height from the top)
        crop_height = int(height * 0.30)
        # Define crop box: (left, top, right, bottom)
        crop_box = (0, crop_height, width, height)

        # Crop the image
        cropped_imagem = imagem.crop(crop_box)

        # Save cropped image to output directory
        output_path = os.path.join(output_directory, os.path.basename(img_path))
        cropped_imagem.save(output_path, "PNG")
        print(f"Saved cropped image: {os.path.basename(output_path)}")

        # Display cropped image if it's one of the 5 selected
        if img_path in display_images:
            print(f"Displaying cropped image: {os.path.basename(img_path)}")
            display(cropped_imagem)

    except Exception as e:
        print(f"Failed to process {os.path.basename(img_path)}: {e}")
```
