```python
# Importing required library for CSV processing
import pandas as pd

# Loading the CSV file from the specified path
file_path = r'D:\iniciacao_cientifica\iniciacao-cientifica\data\Data_Entry_Undergrad_Final.csv'
df = pd.read_csv(file_path)

# Counter for generating sequential IDs starting from 1
counter = 1

# Iterating through the dataframe sequentially
for index, row in df.iterrows():
    # Checking if the current row has 'No Findings' in 'Finding Labels'
    if row['Finding Labels'] == 'No Findings':
        # Assigning formatted ID with leading zeros (e.g., 0001) to Patient ID column
        df.at[index, 'Patient ID'] = f"{counter:04d}"
        counter += 1
    # Stopping when the first 'TB' is encountered
    elif row['Finding Labels'] == 'TB':
        break

# Saving the modified dataframe back to the original CSV file
df.to_csv(file_path, index=False)

# Printing confirmation
print(f"CSV file at {file_path} has been updated with sequential Patient IDs for 'No Findings' rows until the first 'TB'.")
```
