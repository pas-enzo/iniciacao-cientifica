```python
import csv
import os
import pandas as pd
```

```python
# Define file paths
input_dir = r"C:\Users\Windows 10\Documents\IC\projeto-github\iniciacao-cientifica\data"
input_file = os.path.join(input_dir, "Data_Entry_Undergrad.csv")
output_file = os.path.join(input_dir, "Data_Entry_Undergrad_PNG.csv")

# Read the CSV file, ensuring all columns are read
df = pd.read_csv(input_file, keep_default_na=False)

# Verify that 'Other' column exists
if 'Other' not in df.columns:
    print("Warning: 'Other' column not found in the CSV file")
else:
    print("'Other' column found and will be preserved")

# Change extensions in the 'Image Label' column to .png
df['Image Label'] = df['Image Label'].str.replace(r'\.(jpg|jfif)$', '.png', regex=True)

# Save the modified dataframe to a new CSV file, preserving all columns
df.to_csv(output_file, index=False, na_rep='N/A')

print(f"New CSV file saved as: {output_file}")
print(f"Columns in output file: {list(df.columns)}")
```
