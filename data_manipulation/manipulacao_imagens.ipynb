```python
import cv2
import os
import glob
from collections import Counter
import shutil
```

```python
# Define the directory path
input_dir = r"C:\Users\Windows 10\Documents\IC\Imagens\Imagens Oficiais\Entry_Copy_2"

# Get all files in the directory
all_files = os.listdir(input_dir)

# Filter for common image extensions and count them
image_extensions = []
for file in all_files:
    if os.path.isfile(os.path.join(input_dir, file)):
        ext = os.path.splitext(file)[1].lower()
        if ext:  # Only include files with an extension
            image_extensions.append(ext)

# Count the occurrences of each extension
extension_counts = Counter(image_extensions)

# Print the results
print(f"Total files with extensions: {len(image_extensions)}")
print("\nImage formats found in the directory:")
for ext, count in extension_counts.items():
    print(f"{ext}: {count} files")

# Check for files without extensions or unexpected formats
no_extension = [f for f in all_files if not os.path.splitext(f)[1]]
if no_extension:
    print("\nFiles without extensions:")
    for f in no_extension:
        print(f)

# Verify total file count matches expected
print(f"\nExpected total images: 1072")
print(f"Actual files processed: {len(all_files)}")
```

```python
# Define the directory paths
input_dir = r"C:\Users\Windows 10\Documents\IC\Imagens\Imagens Oficiais\Entry_Copy_2"
output_dir = r"C:\Users\Windows 10\Documents\IC\Imagens\Imagens Oficiais\Entry_PNG"

# Create output directory if it doesn't exist
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# Get all .jpg, .jfif, and .png files
image_files = glob.glob(os.path.join(input_dir, "*.jpg")) + glob.glob(os.path.join(input_dir, "*.jfif")) + glob.glob(os.path.join(input_dir, "*.png"))

# Process each image
for image_path in image_files:
    try:
        # Get the filename and extension
        filename, ext = os.path.splitext(os.path.basename(image_path))
        output_path = os.path.join(output_dir, f"{filename}.png")

        if ext.lower() == ".png":
            # Copy PNG files directly
            shutil.copy2(image_path, output_path)
            print(f"Copied {image_path} to {output_path}")
        else:
            # Read and convert .jpg or .jfif to PNG
            img = cv2.imread(image_path)
            if img is None:
                print(f"Failed to read {image_path}")
                continue
            
            # Save as PNG
            cv2.imwrite(output_path, img)
            print(f"Converted {image_path} to {output_path}")
        
    except Exception as e:
        print(f"Error processing {image_path}: {str(e)}")

print("Processing complete!")
```
