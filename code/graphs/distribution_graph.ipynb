```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# --- Dados do seu Projeto de Machine Learning ---
# Estes dados são baseados no output que você forneceu.

# Distribuição de Rótulos do Conjunto de Treino (Imagens)
train_data = {'is_tb': [1] * 506 + [0] * 351}
train_df = pd.DataFrame(train_data)

# Distribuição de Rótulos do Conjunto de Validação (Imagens)
val_data = {'is_tb': [1] * 63 + [0] * 44}
val_df = pd.DataFrame(val_data)

# Distribuição de Rótulos do Conjunto de Teste (Imagens)
test_data = {'is_tb': [1] * 64 + [0] * 44}
test_df = pd.DataFrame(test_data)


# --- Código de Visualização (em Português) ---
print("Visualizando a distribuição de classes para os conjuntos de treino, validação e teste:")
# Criar figura com 3 subplots
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
tb_categories = ['No TB', 'TB']
colors = ['#FF6384', '#36A2EB']

# Gráfico para o Conjunto de Treino
train_counts = train_df['is_tb'].value_counts().sort_index()
train_values = [train_counts.get(0, 0), train_counts.get(1, 0)]
sns.barplot(x=tb_categories, y=train_values, ax=axes[0], palette=colors)
axes[0].set_title(f'Distribuição do Conjunto de Treino (Total: {len(train_df)})')
axes[0].set_ylabel('Número de Casos')
axes[0].set_xlabel('Diagnóstico')
# Adicionar contagens exatas sobre as barras
for p in axes[0].patches:
    axes[0].annotate(f'{int(p.get_height())}', 
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='center', xytext=(0, 10), textcoords='offset points')

# Gráfico para o Conjunto de Validação
val_counts = val_df['is_tb'].value_counts().sort_index()
val_values = [val_counts.get(0, 0), val_counts.get(1, 0)]
sns.barplot(x=tb_categories, y=val_values, ax=axes[1], palette=colors)
axes[1].set_title(f'Distribuição do Conjunto de Validação (Total: {len(val_df)})')
axes[1].set_ylabel('Número de Casos')
axes[1].set_xlabel('Diagnóstico')
# Adicionar contagens exatas sobre as barras
for p in axes[1].patches:
    axes[1].annotate(f'{int(p.get_height())}', 
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='center', xytext=(0, 10), textcoords='offset points')

# Gráfico para o Conjunto de Teste
test_counts = test_df['is_tb'].value_counts().sort_index()
test_values = [test_counts.get(0, 0), test_counts.get(1, 0)]
sns.barplot(x=tb_categories, y=test_values, ax=axes[2], palette=colors)
axes[2].set_title(f'Distribuição do Conjunto de Teste (Total: {len(test_df)})')
axes[2].set_ylabel('Número de Casos')
axes[2].set_xlabel('Diagnóstico')
# Adicionar contagens exatas sobre as barras
for p in axes[2].patches:
    axes[2].annotate(f'{int(p.get_height())}', 
                     (p.get_x() + p.get_width() / 2., p.get_height()),
                     ha='center', va='center', xytext=(0, 10), textcoords='offset points')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# --- Dados de Exemplo (substitua pelos seus dados reais) ---
# A matriz de confusão (cm) e a fase (phase) viriam do cálculo no seu código.
# Usei os valores da sua imagem para que este trecho funcione de forma independente.
cm = np.array([[43, 1], 
               [1, 63]])
phase = "Teste"


# --- Código do Gráfico em Português ---
plt.figure(figsize=(6, 5))
sns.heatmap(cm, 
            annot=True,        # Anotar os números em cada célula
            fmt='d',           # Formatar os números como inteiros
            cmap='Blues',      # Esquema de cores
            cbar=False,        # Não mostrar a barra de cores
            xticklabels=['Negativo Previsto', 'Positivo Previsto'],
            yticklabels=['Negativo Real', 'Positivo Real'])

# Adicionar rótulos e título em português
plt.xlabel('Rótulo Previsto')
plt.ylabel('Rótulo Verdadeiro')
plt.title(f'Matriz de Confusão para o Conjunto de {phase}')

plt.show()
```
